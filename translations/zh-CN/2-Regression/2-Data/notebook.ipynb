```python
import pandas as pd

pumpkins = pd.read_csv("../data/US-pumpkins.csv")
pumpkins.head()
```

```python
print(pumpkins.shape)
```

```python
pumpkins.isnull().sum()
```

```python
pumpkins_clean = pumpkins.drop(columns=['Grade'])

print(pumpkins_clean.shape)
print(pumpkins.shape)
```

```python
print(pumpkins_clean['Package'].value_counts())
```

```python
pumpkins_half_bushel = pumpkins_clean[
    pumpkins_clean['Package'] == '1/2 bushel cartons'
    ]

print(pumpkins_half_bushel.shape)
pumpkins_half_bushel.head()
```

```python
pumpkins_half_bushel = pumpkins_half_bushel.copy()
pumpkins_half_bushel['Average Price'] = (pumpkins_half_bushel['Low Price'] + pumpkins_half_bushel['High Price'])/ 2
pumpkins_half_bushel.head()
```

```python
import matplotlib.pyplot as plt

pumpkins_half_bushel['Average Price'].hist(bins=20)
```

```python
pumpkins_half_bushel['Average Price'].describe()
```

```python
print(pumpkins_half_bushel['Date'].dtype)
pumpkins_half_bushel['Date'] = pd.to_datetime(pumpkins_half_bushel['Date'])
pumpkins_half_bushel['Month'] = pumpkins_half_bushel['Date'].dt.month
pumpkins_half_bushel.head()
```

```python
monthly_prices = pumpkins_half_bushel.groupby('Month')['Average Price'].mean()

monthly_prices.plot(kind='line', marker='o')
```

```python
pumpkins_half_bushel.groupby('Month')['Average Price'].agg(['mean', 'count'])
```

```python
import seaborn as sns

sns.boxplot(data=pumpkins_half_bushel, x='Month', y='Average Price')
```

```python
pumpkins_model_data = pumpkins_half_bushel[['Month', 'Average Price']].copy()
pumpkins_model_data.head()
pumpkins_model_data.isnull().sum()
```

```python
pumpkins_standardized = pumpkins_clean[
    pumpkins_clean['Package'].isin(['1/2 bushel cartons', '1 1/9 bushel cartons'])
].copy()

print(pumpkins_standardized.shape)
print(pumpkins_standardized['Package'].value_counts())
```

```python
pumpkins_standardized['Bushel Size'] = pumpkins_standardized['Package'].map({
    '1/2 bushel cartons': 0.5,
    '1 1/9 bushel cartons': 10/9
})
pumpkins_standardized['Average Price'] = (pumpkins_standardized['Low Price'] + pumpkins_standardized['High Price']) / 2
pumpkins_standardized['Price per Bushel'] = pumpkins_standardized['Average Price'] / pumpkins_standardized['Bushel Size']

pumpkins_standardized[['Package', 'Bushel Size', 'Average Price', 'Price per Bushel']].head()
```

### Assignment：Matplotlib 与 Seaborn 可视化比较

在本次南瓜价格数据探索中，我使用了 Pandas 的 `.hist()` 和 `.plot()` 绘制直方图、折线图，并使用 Seaborn 的 `boxplot()` 绘制箱线图。

从个人使用体验来看，我觉得 Matplotlib 相关的绘图方式更容易理解和使用，因为目前我主要需要完成简单的数据展示。不过，本次 `.hist()` 和 `.plot()` 实际上是 Pandas 的绘图接口，底层使用 Matplotlib。

Seaborn 对统计可视化提供了更方便的接口。例如，使用 `sns.boxplot()` 可以直接比较不同月份的价格分布，观察中位数、四分位距和离群点。

我认为两者各有优势：Matplotlib 适合通用绘图和细节控制，Seaborn 更方便进行统计数据探索。具体选择应该取决于分析任务，而不是简单地认为哪个库更高级。
